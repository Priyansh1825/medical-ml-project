```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

# Medical specific imports
import pydicom
import SimpleITK as sitk

# Set styles
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
%matplotlib inline
```

```python
# Import our custom data loader
import sys
sys.path.append('../src')
from data.loader import DataLoader

# Initialize data loader
loader = DataLoader()

# Generate sample medical data
medical_data = loader.get_sample_medical_data()
print(f"Dataset shape: {medical_data.shape}")
print(f"\nColumns: {medical_data.columns.tolist()}")
medical_data.head()
```

```python
# Basic statistics
print("Basic Statistics:")
print("=" * 50)
print(medical_data.describe())

print("\n\nMissing Values:")
print("=" * 50)
print(medical_data.isnull().sum())
```

```python
# Visualizations
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Age distribution
axes[0, 0].hist(medical_data['age'], bins=20, edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Age Distribution')
axes[0, 0].set_xlabel('Age')
axes[0, 0].set_ylabel('Frequency')

# Gender distribution
gender_counts = medical_data['gender'].value_counts()
axes[0, 1].pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%')
axes[0, 1].set_title('Gender Distribution')

# Blood pressure scatter
axes[1, 0].scatter(medical_data['blood_pressure_sys'], medical_data['blood_pressure_dia'], 
                   c=medical_data['has_disease'], cmap='coolwarm', alpha=0.6)
axes[1, 0].set_title('Blood Pressure: Systolic vs Diastolic')
axes[1, 0].set_xlabel('Systolic BP')
axes[1, 0].set_ylabel('Diastolic BP')

# Disease prevalence by cholesterol
disease_by_chol = medical_data.groupby('cholesterol')['has_disease'].mean().sort_values()
axes[1, 1].bar(disease_by_chol.index, disease_by_chol.values)
axes[1, 1].set_title('Disease Prevalence by Cholesterol Level')
axes[1, 1].set_xlabel('Cholesterol Level')
axes[1, 1].set_ylabel('Disease Rate')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
numeric_cols = medical_data.select_dtypes(include=[np.number]).columns
correlation_matrix = medical_data[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix of Medical Features')
plt.tight_layout()
plt.show()
```
